Esse código está preparando e organizando os dados da tabela exp_final pela data e pelo material para futuramente calcular o trabalho de estoque.

```python
import pandas as pd
exp_final['data'] = pd.to_datetime(exp_final['data'])
exp_final.sort_values(by=['cod_material', 'data'], inplace=True)
```

Agora, vamos calcular o movimento de estoque diário (entrada and saida) para cada material e depois o saldo de estoque acumulado.

```python
import pandas as pd
estoque_df = pd.DataFrame()

for material_code in exp_final['cod_material'].unique():
    material_df = exp_final[exp_final['cod_material'] == material_code].copy()

    # Get unique dates for this material
    min_date = material_df['data'].min()
    max_date = material_df['data'].max()
    full_date_range = pd.date_range(start=min_date, end=max_date, freq='D')

    # Create a daily summary for this material
    daily_summary = pd.DataFrame({'data': full_date_range, 'cod_material': material_code})

    # Calculate daily entries and exits
    entradas_dia = material_df[material_df['tipo'] == 'ENTRADA'].groupby('data')['quantidade'].sum().reindex(full_date_range, fill_value=0)
    saidas_dia = material_df[material_df['tipo'] == 'SAIDA'].groupby('data')['quantidade'].sum().reindex(full_date_range, fill_value=0)

    daily_summary['entradas_dia'] = entradas_dia.values
    daily_summary['saidas_dia'] = saidas_dia.values

    # Calculate accumulated stock
    initial_stock_row = material_df[(material_df['data'] == min_date) & (material_df['observacao'] == 'Saldo inicial')]
    if not initial_stock_row.empty:
        current_stock = initial_stock_row['quantidade'].sum() # Sum if multiple initial entries on the same day
    else:
        current_stock = 0

    accumulated_stock = []
    for index, row in daily_summary.iterrows():
        current_stock += row['entradas_dia'] - row['saidas_dia']
        accumulated_stock.append(current_stock)

    daily_summary['saldo_estoque_acumulado'] = accumulated_stock

    # Merge material details (lead_time_dias, consumo_medio_dia, lote_minimo, custo_unitario, descricao)
    material_info = material_df[['cod_material', 'descricao', 'lead_time_dias', 'consumo_medio_dia', 'lote_minimo', 'custo_unitario']].drop_duplicates()
    daily_summary = pd.merge(daily_summary, material_info, on='cod_material', how='left')

    estoque_df = pd.concat([estoque_df, daily_summary])

estoque_df.reset_index(drop=True, inplace=True)
display(estoque_df.head())
```

Agora, vamos calcular o Estoque de Segurança, o Ponto de Pedido (PP), o Lote de Reposição para cada material. Vamos identificar também os dias no qual o estoque cai para um número menor do que o ponto de pedido ou fica em zero/negativo.

```python
import pandas as pd
# Calculate Safety Stock (Estoque de Segurança)
# Assuming Safety Stock = Lead Time (days) * Average Daily Consumption
estoque_df['estoque_seguranca'] = estoque_df['lead_time_dias'] * estoque_df['consumo_medio_dia']

# Calculate Reorder Point (Ponto de Pedido)
# Assuming Reorder Point = (Lead Time (days) * Average Daily Consumption) + Safety Stock
estoque_df['ponto_pedido'] = (estoque_df['lead_time_dias'] * estoque_df['consumo_medio_dia']) + estoque_df['estoque_seguranca']

# Lote de Reposição (Reorder Lot) is the lote_minimo
estoque_df['lote_reposicao'] = estoque_df['lote_minimo']

# Identify days when stock fell below the Reorder Point
estoque_df['abaixo_pp'] = estoque_df['saldo_estoque_acumulado'] < estoque_df['ponto_pedido']

# Identify days when stock was zero or negative
estoque_df['estoque_zerado_negativo'] = estoque_df['saldo_estoque_acumulado'] <= 0

display(estoque_df.head(10))
```
